```python
# ==========================================
# Task 1: Data Cleaning & Preprocessing
# Titanic Dataset
# ==========================================

# Import Libraries
import pandas as pd
import numpy as np

# ------------------------------------------
# Load Dataset
# ------------------------------------------

# Kaggle Titanic dataset path
df = pd.read_csv('/kaggle/input/datasets/yasserh/titanic-dataset/Titanic-Dataset.csv')

print("Original Dataset Shape:", df.shape)
print("\nFirst 5 Rows:")
print(df.head())
```

```python

# ------------------------------------------
# Basic Information
# ------------------------------------------

print("\nDataset Information:")
print(df.info())

print("\nMissing Values:")
print(df.isnull().sum())
```

```python

# ------------------------------------------
# Remove Duplicate Rows
# ------------------------------------------

duplicates_before = df.duplicated().sum()
print(f"\nDuplicate Rows Before Cleaning: {duplicates_before}")

df = df.drop_duplicates()

duplicates_after = df.duplicated().sum()
print(f"Duplicate Rows After Cleaning: {duplicates_after}")
```

```python

# ------------------------------------------
# Handle Missing Values
# ------------------------------------------

# Age -> Fill with Median
df['Age'] = df['Age'].fillna(df['Age'].median())

# Embarked -> Fill with Mode
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# Cabin -> Too many missing values
# Fill with 'Unknown'
df['Cabin'] = df['Cabin'].fillna('Unknown')

print("\nMissing Values After Handling:")
print(df.isnull().sum())
```

```python

# ------------------------------------------
# Convert Data Types
# ------------------------------------------

# Convert Survived to Category
df['Survived'] = df['Survived'].astype('category')

# Convert Pclass to Category
df['Pclass'] = df['Pclass'].astype('category')

# Convert Sex to Category
df['Sex'] = df['Sex'].astype('category')

# Convert Embarked to Category
df['Embarked'] = df['Embarked'].astype('category')
```

```python

# ------------------------------------------
# Rename Columns
# ------------------------------------------

df = df.rename(columns={
    'PassengerId': 'Passenger_ID',
    'Survived': 'Survived_Status',
    'Pclass': 'Passenger_Class',
    'Name': 'Passenger_Name',
    'Sex': 'Gender',
    'Age': 'Age_Years',
    'SibSp': 'Siblings_Spouses',
    'Parch': 'Parents_Children',
    'Ticket': 'Ticket_Number',
    'Fare': 'Ticket_Fare',
    'Cabin': 'Cabin_Number',
    'Embarked': 'Embarkation_Port'
})
```

```python

# ------------------------------------------
# Verify Changes
# ------------------------------------------

print("\nCleaned Dataset Information:")
print(df.info())

print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 5 Rows After Cleaning:")
print(df.head())
```

```python

# ------------------------------------------
# Save Cleaned Dataset
# ------------------------------------------

output_path = '/kaggle/working/titanic_cleaned.csv'
df.to_csv(output_path, index=False)

print(f"\nCleaned dataset saved successfully at:")
print(output_path)

print("\nFinal Dataset Shape:", df.shape)
```
